```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# zapewnienie powtarzalności
np.random.seed(2025)

# tworzenie danych
x_train = np.arange(0, 12)
x_test = (np.arange(0, 12) + 0.5)[:-1]
random_noise_train = np.random.normal(loc=0, scale=10, size=len(x_train))
random_noise_test = np.random.normal(loc=0, scale=10, size=len(x_test))
y_train = x_train**2 + random_noise_train
y_test = x_test**2 + random_noise_test

# rysowanie wykresu
plt.scatter(x_train, y_train, color = 'steelblue')
plt.scatter(x_test, y_test, color = 'coral')
plt.show()
```

```python
# uczenie modelu drugiego stopnia
coefficient = np.polyfit(x_train, y_train, deg=2)

# ewaluacja modelu
y_fit = np.polyval(coefficient, x_train)
y_fit_test = np.polyval(coefficient, x_test)

print("Współczynniki:", coefficient)
```

```python
# rysowanie wykresu z dopasowanym modelem
plt.scatter(x_train, y_train, color = 'steelblue')
plt.scatter(x_test, y_test, color = 'coral')
plt.plot(x_train, y_fit, color="red", label="Model")
plt.show()
```

```python
# wartości błędu
red_train = np.sqrt(np.mean((y_train - y_fit)**2))
red_test = np.sqrt(np.mean((y_test - y_fit_test)**2))

print("Wartość błędu na zbiorze treningowym: ", round(red_train, 2))
print("Wartość błędu na zbiorze testowym: ", round(red_test, 2))
```

```python
# uczenie modelu dwunastego stopnia
coefficient = np.polyfit(x_train, y_train, deg=1)

# ewaluacja modelu
y_fit_2 = np.polyval(coefficient, x_train)
y_fit_test_2 = np.polyval(coefficient, x_test)

print("Współczynniki:", coefficient)
```

```python
# rysowanie wykresu z przetrenowanym modelem
x_plot = np.linspace(0, 11, 100)
y_plot = np.polyval(coefficient, x_plot)
plt.scatter(x_train, y_train, color = 'steelblue')
plt.scatter(x_test, y_test, color = 'coral')
plt.plot(x_plot, y_plot, color="blue", label="Model")
plt.show()
```

```python
# wartości błędu
blue_train = np.sqrt(np.mean((y_train - y_fit_2)**2))
blue_test = np.sqrt(np.mean((y_test - y_fit_test_2)**2))

print("Wartość błędu na zbiorze treningowym: ", round(blue_train, 2))
print("Wartość błędu na zbiorze testowym: ", round(blue_test, 2))
```

```python
# wspólny wykres
plt.scatter(x_train, y_train, color = 'steelblue')
plt.scatter(x_test, y_test, color = 'coral')
plt.plot(x_train, y_fit, color="red", label="Model")
plt.plot(x_plot, y_plot, color="blue", label="Model")
plt.show()

print("Czerwony:")
print("Wartość błędu na zbiorze treningowym: ", round(red_train, 2))
print("Wartość błędu na zbiorze testowym: ", round(red_test, 2))
print("\n")
print("Niebieski:")
print("Wartość błędu na zbiorze treningowym: ", round(blue_train, 2))
print("Wartość błędu na zbiorze testowym: ", round(blue_test, 2))
```
