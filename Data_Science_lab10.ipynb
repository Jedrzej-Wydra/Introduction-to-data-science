```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd
import numpy as np
import random

from matplotlib.colors import ListedColormap
from tqdm import tqdm
```

```python
sf_step = 0.01
lower_limit = -10
upper_limit = 10 + sf_step

x = np.arange(lower_limit, upper_limit, sf_step)
y = np.arange(lower_limit, upper_limit, sf_step)

X, Y = np.meshgrid(x, y)
Z = 0.1 * (X**2 + Y**2)

Z[Z > 10] = np.nan
```

```python
turbo = px.colors.sequential.Turbo

green_turbo = turbo[3:]

fig = go.Figure(
    data=go.Surface(
        x=x, y=y, z=Z,
        surfacecolor=Z,
        colorscale=green_turbo,    # "Turbo", "Plasma", "Cividis", "Viridis"
        cmin=0,
        cmax=10,
        colorbar=dict(title="Z"),
        showscale=False
    )
)

fig.update_layout(
    scene=dict(
        xaxis=dict(showgrid=False,
                   zeroline=False,
                   showticklabels=False,
                   showbackground=False,
                   title=""),
        yaxis=dict(showgrid=False,
                   zeroline=False,
                   showticklabels=False,
                   showbackground=False,
                   title=""),
        zaxis=dict(showgrid=False,
                   zeroline=False,
                   showticklabels=False,
                   showbackground=False,
                   title=""),
    )
)

#fig.update_traces(opacity=0.9)

fig.show()
```

**Zadanie 2.**

Za pomocą algorytmu gradientowego proszę zasymulować działanie siły grawitacji. Tzn. proszę wylosować punkt w górnych częściach powierzchni, a następnie za pomocą algorytmu znaleźć położenia tego punktu w kolejnych punktach czasowych jak gdyby toczył się zgodnie z siłą grawitacji. Proszę pominąć przyspieszenie.

```python
X0, Y0 = np.random.uniform(6.5,7,2)
px_first = [X0]
py_first = [Y0]
pz_first = [0.1*(X0**2 + Y0**2) + 0.1]
fig = go.Figure(
    data=go.Surface(
        x=x, y=y, z=Z,
        surfacecolor=Z,
        colorscale=green_turbo,
        cmin=0,
        cmax=10,
        colorbar=dict(title="Z"),
        showscale=False
    )
)

fig.add_trace(go.Scatter3d(
    x=px_first, y=py_first, z=pz_first,
    mode="markers",
    marker=dict(size=4, color="black"),
    name="pierwszy"
))

fig.update_layout(
    scene=dict(
        xaxis=dict(showgrid=False,
                   zeroline=False,
                   showticklabels=False,
                   showbackground=False,
                   title=""),
        yaxis=dict(showgrid=False,
                   zeroline=False,
                   showticklabels=False,
                   showbackground=False,
                   title=""),
        zaxis=dict(showgrid=False,
                   zeroline=False,
                   showticklabels=False,
                   showbackground=False,
                   title=""),
    ),
    showlegend = False
)

#fig.update_traces(opacity=0.9)

fig.show()
```

```python
gravity = [[X0, Y0]]

def gradient(lista, lr = 0.1, epsilon = 0.001):
  cond = True
  while cond:
    X, Y = lista[-1]
    loss1 = 0.1 * (X**2 + Y**2)
    X, Y = X - lr*0.2*X, Y - lr*0.2*Y
    loss2 = 0.1 * (X**2 + Y**2)
    lista.append([X, Y])
    cond = abs(loss1 - loss2) > epsilon
  return lista, len(lista)

coords, gens = gradient(gravity)

pts = np.array(coords[::6], dtype=float)
px, py = pts[:, 0], pts[:, 1]
pz = 0.1 * (px**2 + py**2) + 0.2

px_all, py_all, pz_all = px[1:-1], py[1:-1], pz[1:-1]
px_first, py_first, pz_first = [px[0]], [py[0]], [pz[0]]
px_last, py_last, pz_last = px[-1:], py[-1:], pz[-1:]

fig = go.Figure(
    data=go.Surface(
        x=x, y=y, z=Z,
        surfacecolor=Z,
        colorscale=green_turbo,
        cmin=0,
        cmax=10,
        colorbar=dict(title="Z"),
        showscale=False
    )
)

fig.add_trace(go.Scatter3d(
    x=px_first, y=py_first, z=pz_first,
    mode="markers",
    marker=dict(size=4, color="black"),
    name="pierwszy"
))

fig.add_trace(go.Scatter3d(
    x=px_all, y=py_all, z=pz_all,
    mode="markers",
    marker=dict(size=4, color="gray"),
    name="punkty"
))

fig.add_trace(go.Scatter3d(
    x=px_last, y=py_last, z=pz_last,
    mode="markers",
    marker=dict(size=6, color="red"),
    name="ostatni"
))

fig.update_layout(
    scene=dict(
        xaxis=dict(showgrid=False,
                   zeroline=False,
                   showticklabels=False,
                   showbackground=False,
                   title=""),
        yaxis=dict(showgrid=False,
                   zeroline=False,
                   showticklabels=False,
                   showbackground=False,
                   title=""),
        zaxis=dict(showgrid=False,
                   zeroline=False,
                   showticklabels=False,
                   showbackground=False,
                   title=""),
    ),
    showlegend = False
)

fig.update_traces(opacity=0.9)

fig.show()
```

**Zadanie 4.**

Proszę zaimplementować algorytm genetyczny do szukania minimum.

```python
population = pd.DataFrame({
    'X':[],
    'Y':[],
    'id':[],
    'd_id':[],
    'm_id':[],
    'gen':[],
    'age':[],
    'prob':[],
    'die':[]
})

init_size = 30
lim1 = 6
lim2 = 7

for i in range(init_size):
  row = len(population)
  xx = np.round(np.random.uniform(lim1, lim2), 3)
  yy = np.round(np.random.uniform(lim1, lim2), 3)
  p = 0.01 + 0.01*(xx**2 + yy**2)/2
  if p > 1:
    p = 1
  population.loc[row] = {'X': xx,
                     'Y': yy,
                     'id': i,
                      'd_id': i,
                     'm_id': i,
                     'gen': 0,
                     'age': 0,
                     'prob': p,
                      'die': 0}
```

```python
def reproduce(population, maxkid = 5, mutation = 1):

  died_population = population[population['die'] == 1]
  living_population = population[population['die'] == 0]
  pop = list(living_population.id)
  pairs = []
  marrige = []
  for i in range(2):
    new_pop = []
    random.shuffle(pop)
    if len(pop) % 2 == 0:
      pairs = np.array(pop).reshape(-1, 2)
    else:
      pairs = np.array(pop[:-1]).reshape(-1, 2)
      new_pop.append(pop[-1])

    for i in range(len(pairs)):
      set1 = {living_population.d_id[pairs[i,0]], living_population.m_id[pairs[i,0]]}
      set2 = {living_population.d_id[pairs[i,1]], living_population.m_id[pairs[i,1]]}
      pm = set1.isdisjoint(set2)
      if pm:
        marrige.append(pairs[i,:])
      else:
        new_pop.append(pairs[i,0])
        new_pop.append(pairs[i,1])

    pop = new_pop

  for para in marrige:
    dad = population[population.id == int(para[0])]
    mum = population[population.id == int(para[1])]
    kids = int(np.random.choice(list(range(maxkid))))
    gen = max([population.gen[int(para[0])], population.gen[int(para[1])]])
    for j in range(kids):
      id = len(population)
      gene_mutation = [0, np.random.normal(mutation, 0.3)]
      random.shuffle(gene_mutation)
      gensX = np.random.choice([list(dad.X)[0], list(mum.X)[0]]) + gene_mutation[0]
      gensY = np.random.choice([list(dad.Y)[0], list(mum.Y)[0]]) + gene_mutation[1]
      p = 0.01 + 0.01*(gensX**2 + gensY**2)/2
      if p > 1:
        p = 1
      population.loc[id] = {
          'X':round(gensX,3),
          'Y':round(gensY,3),
          'id':id,
          'd_id': int(para[0]),
          'm_id': int(para[1]),
          'gen': gen + 1,
          'age':0,
          'prob':p,
          'die':0
      }

  return population

def die(population):
    mask = population["die"] == 0

    population.loc[mask, "age"]  = population.loc[mask, "age"] + 1
    population.loc[mask, "prob"] = (population.loc[mask, "prob"] * 1.2).clip(upper=0.9)

    population.loc[mask, "die"] = np.random.binomial(
        n=1,
        p=population.loc[mask, "prob"])

    return population

```

```python
mut0 = 1

for i in tqdm(range(12)):

  mut1 = population[population['gen'] == population['gen'].max()]['prob'].mean()
  mutation = 6*(mut1 - mut0)
  population = reproduce(die(population), maxkid=6, mutation = mutation)
  mut0 = population[population['gen'] == population['gen'].max() - 1]['prob'].mean()

population
```

```python
mean_xy = (
    population
    .groupby("gen", as_index=False)
    .agg(
        mean_X=("X", "mean"),
        mean_Y=("Y", "mean"),
        mean_prob=("prob", "mean")
    )
    .reset_index()
)


mean_xy
```
