# Lepiej zdawać długie, czy krótkie testy?

**Zadanie 1.**

Proszę narysować histogramy i esytmatory gęstości jądrowej dla:
1. próby wylosowanej z rozkładu P;
2. próby powstałej z wyliczenia średnich z prób z rozkładu P.

Jako rozkład P proszę zastosować:
1. Rozkład jednostajny;
2. Rozkład chi-kwadrat;
3. Rozkłąd Weibulla.

Próba średnich powinna mieć sto elementów. Próby z których liczona jest średnia powinny mieć 1, 10, 50 i 100 obserwacji.

Zatem należy otrzymać 24 wykresy.

**Zadanie 2.**

Proszę napisać funkcję tworzącą test.

**Zadanie 3.**

Proszę napisać funkcję, która wybiera poprawnę odpowiedź z prawdopodobieństwem p.

**Zadanie 4.**

Dla liczb pytań: 15, 30, 45, 60, 75, 90, proszę narysować wykresy wynik test vs. prawdopodobieństwo poprawnej odpowiedzi wraz z przedziałami ufności wyliczonymi zgodnie z centralnym twierdzeniem granicznym.

**Zadanie 5.**

Dla prawdopodobieństw: 0.15, 0.3, 0.45, 0.6, 0.75, 0.9, proszę narysować wykresy wynik test vs. liczba pytań wraz z przedziałami ufności wyliczonymi zgodnie z centralnym twierdzeniem granicznym.
