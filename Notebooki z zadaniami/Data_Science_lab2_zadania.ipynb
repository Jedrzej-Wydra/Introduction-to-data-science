## Zadanie 8
Proszę wyznaczyć: **sumę**, **różnicę**, **iloczyn** i **iloraz** macierzy:

$$
A =
\begin{bmatrix}
1 & 4 \\
6 & 3
\end{bmatrix}
$$
oraz
$$
B =
\begin{bmatrix}
2 & 8 \\
7 & 0
\end{bmatrix}
$$

Przez **iloraz macierzy** $A$ i $B$ należy tutaj rozumieć **iloczyn macierzy $A$** i **macierzy odwrotnej do $B$**, czyli:

$$
A \cdot B^{-1}
$$
