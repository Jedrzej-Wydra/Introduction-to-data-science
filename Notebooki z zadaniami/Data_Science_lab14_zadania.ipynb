# Łańcuchy Markova wytrenowane na treści AI Act

**Zadanie 1.**

Proszę załadować do środowiska Pythona tekst unijnego Rozporzdzenia regulującego sztuczną inteligencję (tzw. AI Act). Rozporządzenie jest dostępne w serwisie EUR-Lex.

**Zadanine 2.**

Proszę wytrenować łańcuch Markova.

**Zadanie 3.**

Proszę zdefiniować funkcję do generowania tekstu.

**Zadanie 4.**

Proszę wygenerować tekst za pomocą wytrenowanych łańcuchów Markova. Proszę uwzględnić poniższe instrukcje:
1. Jeśli użytkownik nie poda początkowego słowa, niech funkcja generująca tekst losowo wybiera pierwsze słowo na podstawie prawdopodobieństwa wyznaczonego na podstawie częstości słów w tekście rozporządzenia;
2. Niech kolejne słowo będzie generowane na podstawie trzech poprzednich;
3. Jeśli liczba poprzednich słów jest mniejsza od trzech, to niech kolejne jest generowane na podstawie dwóch poprzednich;
4. Jeśli liczba poprzednich słów jest mniejsza od dwóch, to niech kolejne jest generowane na podstawie jednego poprzedniego słowa;
5. Niech funkcja przyjmuje jako argument liczbę naturalną określającą jak długi ma być ciąg wygenerowanych słów.

**Zadanie 5.**

Proszę wyświetlić fragment macierzy przejścia.

**Zadanie 6.**

Proszę obliczyć wymiar pełnej macierzy przejścia.
