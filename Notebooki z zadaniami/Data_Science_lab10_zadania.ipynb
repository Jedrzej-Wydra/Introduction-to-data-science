# Algorytm Genetyczny

**Zadanie 1.**

Proszę narysować powierzchnię w trzech wymiarach o jednym minimum.

**Zadanie 2.**

Za pomocą algorytmu gradientowego proszę zasymulować działanie siły grawitacji. Tzn. proszę wylosować punkt w górnych częściach powierzchni, a następnie za pomocą algorytmu znaleźć położenia tego punktu w kolejnych punktach czasowych jak gdyby toczył się zgodnie z siłą grawitacji. Proszę pominąć przyspieszenie.

**Zadanie 3.**

Proszę narysować powierzchnię jako mapę ciepła.

**Zadanie 4.**

Proszę zaimplementować algorytm genetyczny do szukania minimum.

# Propozycje projektów

## Propozycja 1

Proszę zoptymalizować kod przedstawiony w niniejszym notatniku poprzez przyspieszenie algorytmu genetycznego. Podczas prezentacji należy wyjaśnić jak działa algorytm genetyczny oraz co zostało zrobione aby przyspieszyć jego działanie.

## Propozycja 2

Proszę wykonać analogiczny projekt jak w niniejszym notatniku, ale na powierzchni z dwoma minimami.

## Propozycja 3

Proszę wyjaśnić jak działa algorytm genetyczny, w tym proszę pokazać dlaczego crossover i mutacje są istotne oraz dlaczego wpływają na algorytm. Proszę przygotować symulacje algorytmu genetycznego, algorytmu bez crossover, algorytmu bez mutacji. Następnie porównać czas wykonywania i efekty.
