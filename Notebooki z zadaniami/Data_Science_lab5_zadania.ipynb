# Ile pracy potrzeba, żeby nauczyć komputer wykonywania prostej czynności?

**Zadanie 0.**

Proszę przygotować czarno-biały obrazek, który będzie wykorzystywany w dalszej części projektu.

**Zadanie 1.**

Proszę załadować przygotowany obrazek do środowiska Pythona, a następnie przekształcić go w macierz binarną. Warto zmniejszyć obrazek, aby przyspieszyć obliczenia.

**Zadanie 2.**

Proszę zmienić macierz w ramkę danych, w której w kolumnie pierwszej są współrzędne punktów na osi OX, a w kolumnie drugiej współrzędne punktów na osi OY. Następnie proszę przekształcić współrzędne tak, aby środek obrazka znajdował się w środku układu współrzędnych.

Podpowiedź: Niezbędne jest dodatkowe przekształcenie współrzędnych na osi OY, aby obrazek miał poprawną orientację.

**Zadanie 3.**

Proszę:
1. narysować obrazek jako scatter plot;
2. przygotować funkcję macierzy obrotu;
3. narysować obrazek obrócony o 30 stopni jako scatter plot.

**Zadanie 4.**

Proszę przygotować dane, które pozwolą nauczyć model liniowy obracać obrazek o pewien kąt.

**Zadanie 5.**

Proszę nauczyć model liniowy obracania obrazka o pewien kąt, a następnie obrócić obrazek i narysować jako scatter plot.

## Propozycja projektu

Proszę zaprezentować rozwiązania powyższych zadań oraz:
1. przedstawić rozumowanie analityczne, na podstawie którego zostały przygotowane dane do uczenia modelu;
2. przygotować funkcję, która obraca obrazek o wybrany kąt za pomocą uczenia maszynowego.

Funkcja z punktu drugiego może być funkcją, która obraza o dowolny kąt lub o kąt z wcześniej zdefiniowanej listy. Oprócz zadania trzeciego, proszę nigdzie nie wykorzystywać macierzy obrotu.
