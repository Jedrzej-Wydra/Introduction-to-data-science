# Podział danych na zbiór treningowy i zbiór testowy

**Zadanie 1.**

Proszę napisać funkcję generującą zbiór danych losowych z trendem liniowym o 12 obserwacjach.

**Zadanie 2.**

Za pomocą funkcji z zadania 1. proszę utworzyć zbiór danych i dopasować do niego wielomian dwunastego stopnia.

**Zadanie 3.**

Do danych utworzonych w zadaniu 2. proszę dopasować model liniowy.

**Zadanie 4.**

Proszę dokonać analizy błędu, tzn. obliczyć średni błąd kwadratowy modelu z zadania 2. i modelu z zadania 3.

**Zadanie 5.**

Za pomocą funkcji z zadania 1 proszę wygenerować nowy zbiór danych i sprawdzić jakie błędy generują na nim modele z zdań 2. i 3.

**Zadanie 6.**

Proszę napisać funkcję generującą zbiór danych losowych z trendem kwadratowym.

**Zadanie 7.**

Za pomocą funkcji z zdania 6. proszę utworzyć zbiór danych i dopasować do niego model liniowy.

**Zadanie 8.**

Do danych z zadania 7. proszę dopasować model kwadratowy.

**Zadanie 9.**

Proszę przeprowadzić analizę błędu modeli z zadań 7. i 8. analogicznie jak w zadaniu 4.

**Zadanie 10.**

Za pomocą funkcji z zadania 6. proszę wygenerować nowy zbiór danych i analogicznie jak w zadaniu 5. sprawdzić jakie błędy generują na nim modele z zdań 7. i 8.
