4. zwracającą informacje o danych przechowywanych przez obiekt.

````python
class profesions_class:
  def __init__(self, dict_arg):
    self.ans = dict_arg

  def sorted_names(self):
    self.result = sorted([i[0].split(' ')[1] for i in self.ans.items()])
    return self.result

  def numbered_profesions(self):
    self.result = []
    self.helper_list = list(self.ans.values())
    for i in range(len(self.helper_list)):
      self.result.append((i + 1, self.helper_list[i]))
    return self.result

  def unordered_names(self):
    self.helper_list = [i[0].split(' ')[0] for i in self.ans.items()]
    self.result = set(self.helper_list)
    return self.result

  def __str__(self):
    lines = []
    for name, job in self.ans.items():
      lines.append(f"imię i nazwisko: {name}\nzawód: {job}\n")
    return "\n".join(lines)
````

## Zadanie 13
Proszę zdefiniować pythonową funkcję, która rozwiązuje równianie kwadratowe postaci $ax^2 + bx + c = 0$ bez używania Numpy.

````python
def solve_quadratic(a, b, c):
    if a == 0 and b == 0 and c == 0:
        return "Nieskończenie wiele rozwiązań"
    if a == 0 and b != 0:
        return (-c/b, )
    if a == 0 and b == 0 and c != 0:
        return "Brak rozwiązań"
    delta = b**2 - 4*a*c
    if delta < 0:
        return "Brak rozwiązań rzeczywistych"
    sqrt_delta = delta**0.5
    if delta == 0:
        x0 = -b/(2*a)
        return (x0, )
    else:
        x1 = (-sqrt_delta-b)/(2*a)
        x2 = (sqrt_delta-b)/(2*a)
        if x1 < x2:
            return (x1, x2)
        else:
            return (x2, x1)

solve_quadratic(1, 1, -1)
````

## Zadanie 26
Proszę napisać funkcję wyznaczającą wyznacznik macierzy kwadratowej, korzystając tylko z funkcjonalności bazowego Pythona.

````python
def mat_det(mat):
  if len(mat) == 1:
    return mat[0][0]
  res = 0
  for i in range(len(mat[0])):
    if i % 2 == 0:
      s = 1
    else:
      s = -1
    mat2 = [j[:i] + j[(i+1):] for j in mat[1:]]
    res += s*mat[0][i]*mat_det(mat2)

  return res
````

````python
mat_det([[1,2,4],[2,3,6],[9,3,4]])
````

## Zadanie 31
Proszę przygotować funkcję, która zamieni wszystkie daty w tekście z formatu DD-MM-YYYY na YYYY/MM/DD, za pomocą regex. Na przykład: 12-12-2024 -> 2024/12/12

````python
def date_conversion(text):
  pattern = r"\b(\d{2})-(\d{2})-(\d{4})\b"

  result = re.sub(pattern, r"\3/\2/\1", text)

  return result
````
